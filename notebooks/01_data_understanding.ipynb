```python
from pathlib import Path
import pandas as pd

# Find the project folder
PROJECT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_DIR / "data" / "raw"

# Load the student information table
students = pd.read_csv(DATA_DIR / "studentInfo.csv")

# Display the first five rows
display(students.head())
```

```python
print("Dataset shape:", students.shape)

print("\nFinal outcome counts:")
print(students["final_result"].value_counts())

print("\nMissing values:")
print(students.isnull().sum())
```

```python
# Research check 1: Dataset quality and outcome distribution

print("Dataset dimensions:")
print(f"Students/records: {students.shape[0]:,}")
print(f"Columns: {students.shape[1]}")

print("\nDuplicate rows:")
print(students.duplicated().sum())

print("\nUnique student IDs:")
print(students["id_student"].nunique())

print("\nFinal outcomes with percentages:")
outcome_summary = (
    students["final_result"]
    .value_counts()
    .to_frame(name="count")
)

outcome_summary["percentage"] = (
    outcome_summary["count"] / len(students) * 100
).round(2)

display(outcome_summary)

print("\nMissing values by column:")
display(students.isnull().sum().to_frame(name="missing_values"))
```

```python
# Check for duplicate records and repeated student IDs

print("Exact duplicate rows:", students.duplicated().sum())

print("Total student-course records:", len(students))

print("Unique student IDs:", students["id_student"].nunique())

print("\nRecords per student:")
records_per_student = students["id_student"].value_counts()

print(records_per_student.value_counts().sort_index())
```
